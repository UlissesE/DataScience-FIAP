# Parte I — Regressão Linear Simples

## 6. Conceito

Na **Regressão Linear Simples**, utilizamos apenas uma variável explicativa para estimar uma variável alvo.

Neste exemplo:

- `X = MedInc`
- `y = MedHouseVal`

O modelo busca uma reta que represente a relação entre essas variáveis.

\[
\hat{y} = b_0 + b_1X
\]

Onde:

- `b₀`: intercepto;
- `b₁`: coeficiente angular;
- `X`: variável explicativa;
- `ŷ`: valor previsto.

## 7. Definindo X e y

```python
X = df[['MedInc']]
y = df['MedHouseVal']

print("Formato de X:", X.shape)
print("Formato de y:", y.shape)
```

## 8. Divisão em treino e teste

Vamos utilizar:

- 80% dos dados para treinamento;
- 20% para teste.

O conjunto de teste permite avaliar o modelo em dados que ele não utilizou durante o treinamento.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Treino:", X_train.shape)
print("Teste:", X_test.shape)
```

## 9. Construindo a Regressão Linear Simples

```python
modelo_simples = LinearRegression()

modelo_simples.fit(X_train, y_train)

print("Intercepto:", modelo_simples.intercept_)
print("Coeficiente de MedInc:", modelo_simples.coef_[0])
```

## 10. Interpretando o coeficiente

O coeficiente associado a `MedInc` representa quanto o modelo estima que `MedHouseVal` varia quando `MedInc` aumenta uma unidade.

A interpretação deve sempre considerar as unidades utilizadas pelo dataset.

## 11. Fazendo previsões

```python
y_pred_simples = modelo_simples.predict(X_test)

comparacao_simples = pd.DataFrame({
    'valor_real': y_test.values,
    'valor_previsto': y_pred_simples
})

comparacao_simples.head(10)
```

## 12. Visualizando a reta de regressão

```python
ordem = np.argsort(X_test['MedInc'].values)

plt.figure(figsize=(8, 5))

plt.scatter(
    X_test['MedInc'],
    y_test,
    alpha=0.25,
    label='Valores reais'
)

plt.plot(
    X_test['MedInc'].values[ordem],
    y_pred_simples[ordem],
    linewidth=2,
    label='Reta de regressão'
)

plt.title('Regressão Linear Simples — MedInc')
plt.xlabel('MedInc')
plt.ylabel('MedHouseVal')
plt.legend()
plt.show()
```

## 13. Métricas de avaliação

### MAE — Mean Absolute Error

Representa a média do valor absoluto dos erros.

**Quanto menor, melhor.**

### RMSE — Root Mean Squared Error

Também mede o erro, mas penaliza mais fortemente erros maiores.

**Quanto menor, melhor.**

### R² — Coeficiente de determinação

Indica quanto da variabilidade da variável alvo é explicada pelo modelo.

**Quanto mais próximo de 1, maior o poder explicativo do modelo.**

```python
mae_simples = mean_absolute_error(y_test, y_pred_simples)

rmse_simples = np.sqrt(
    mean_squared_error(y_test, y_pred_simples)
)

r2_simples = r2_score(y_test, y_pred_simples)

print("Regressão Linear Simples")
print(f"MAE:  {mae_simples:.4f}")
print(f"RMSE: {rmse_simples:.4f}")
print(f"R²:   {r2_simples:.4f}")
```

## 14. Valores reais × valores previstos

```python
plt.figure(figsize=(6, 6))

plt.scatter(
    y_test,
    y_pred_simples,
    alpha=0.3
)

plt.xlabel('Valor Real')
plt.ylabel('Valor Previsto')
plt.title('Real × Previsto — Regressão Linear Simples')

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)

plt.show()
```

## 15. Análise de resíduos

O resíduo é a diferença entre o valor observado e o valor previsto:

`resíduo = valor real - valor previsto`

```python
residuos_simples = y_test.values - y_pred_simples

plt.figure(figsize=(8, 5))

plt.scatter(
    y_pred_simples,
    residuos_simples,
    alpha=0.3
)

plt.axhline(0, linewidth=1)

plt.title('Resíduos — Regressão Linear Simples')
plt.xlabel('Valor previsto')
plt.ylabel('Resíduo')
plt.show()
```

```python
plt.figure(figsize=(7, 4))

plt.hist(
    residuos_simples,
    bins=40
)

plt.title('Distribuição dos Resíduos — Modelo Simples')
plt.xlabel('Resíduo')
plt.ylabel('Frequência')
plt.show()
```

# Parte II — Regressão Linear Múltipla

## 16. Conceito

Na **Regressão Linear Múltipla**, utilizamos duas ou mais variáveis explicativas.

\[
\hat{y} = b_0 + b_1X_1 + b_2X_2 + \cdots + b_nX_n
\]

No California Housing, podemos utilizar simultaneamente:

- `MedInc`;
- `HouseAge`;
- `AveRooms`;
- `AveBedrms`;
- `Population`;
- `AveOccup`;
- `Latitude`;
- `Longitude`.

Cada coeficiente representa a relação estimada de uma variável com a variável alvo **considerando as demais variáveis presentes no modelo**.

## 17. Selecionando todas as variáveis explicativas

```python
features = data.feature_names

X_multi = df[features]
y_multi = df['MedHouseVal']

print("Variáveis utilizadas:")
print(features)

X_multi.head()
```

## 18. Dividindo os dados

```python
X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_multi,
    y_multi,
    test_size=0.20,
    random_state=42
)
```

## 19. Construindo a Regressão Linear Múltipla

Para `LinearRegression`, a padronização não é obrigatória para gerar as previsões do modelo linear.

Neste laboratório manteremos os dados em suas unidades originais, pois isso facilita a interpretação dos coeficientes.

```python
modelo_multiplo = LinearRegression()

modelo_multiplo.fit(X_train_m, y_train_m)

y_pred_multiplo = modelo_multiplo.predict(X_test_m)

print("Intercepto:", modelo_multiplo.intercept_)
```

## 20. Coeficientes do modelo múltiplo

```python
coeficientes = pd.DataFrame({
    'variavel': features,
    'coeficiente': modelo_multiplo.coef_
})

coeficientes
```

## 21. Avaliando a Regressão Linear Múltipla

```python
mae_multiplo = mean_absolute_error(
    y_test_m,
    y_pred_multiplo
)

rmse_multiplo = np.sqrt(
    mean_squared_error(
        y_test_m,
        y_pred_multiplo
    )
)

r2_multiplo = r2_score(
    y_test_m,
    y_pred_multiplo
)

print("Regressão Linear Múltipla")
print(f"MAE:  {mae_multiplo:.4f}")
print(f"RMSE: {rmse_multiplo:.4f}")
print(f"R²:   {r2_multiplo:.4f}")
```
